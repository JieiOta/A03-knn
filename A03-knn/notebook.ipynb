```python
# Question 2
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
df = pd.read_csv('/Users/jieiota/Documents/Year 3 Semester 1/DS 3001/A03-knn/data/land_mines.csv')
print(df.describe())
print(df.shape)
print(df.info())
df.head()

X = df[['voltage', 'height', 'soil']]
Y = df['mine_type']

X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.5, random_state=42
)

print( X_train,'\n', X_test)
print( y_train,'\n', y_test)


classification_scaler = MinMaxScaler()
X_train_scaled = classification_scaler.fit_transform(X_train)
X_test_scaled = classification_scaler.transform(X_test)

eval_ks = np.arange(1, 51)
eval_valid = []
for eval_k in eval_ks:
    model = KNeighborsClassifier(n_neighbors= int(eval_k))
    model.fit(X_train_scaled, y_train)
    predictions = model.predict(X_test_scaled)
    eval_valid.append(np.mean(predictions == y_test))

best_k = int(eval_ks[np.argmax(eval_valid)])
knn = KNeighborsClassifier(n_neighbors=best_k)
knn.fit(X_train_scaled, y_train)

# k is selected by going from values 1-50 and testing the accuracy of each one to find the highest.

cm = confusion_matrix(y_test, knn.predict(X_test_scaled))
print(cm)

# With every predictive modeling, errors can still occur and with that I'd advise someone to also be monitoring the outputs
# from these as well if possible as a human verification can help improve success rates. Additionally, in some instances a false
# positive or a false negative may be more detrimental, so training the model to lean further away from one of those could be helpful as well. 

```
